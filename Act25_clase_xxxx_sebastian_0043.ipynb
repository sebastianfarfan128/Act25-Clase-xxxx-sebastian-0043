### 🏨 Descripción del Problema: Clase Hotel
El objetivo es modelar un sistema que permita gestionar la información de establecimientos de alojamiento. Cada hotel es tratado como un objeto único con su propia identidad y categoría dentro de un catálogo turístico.

###1. Definición de Atributos
Los atributos representan las características esenciales que definen a cada hotel:

* nombre: Cadena de texto que identifica legal y comercialmente al establecimiento.

* ubicación: Dirección física o coordenadas que determinan la localización del inmueble.

* estrellas: Valor numérico (normalmente de 1 a 5) que define la categoría y calidad de los servicios.

###2. Definición de Funciones (Métodos)
Las funciones definen las acciones de interacción con la información del hotel:

* leerDatos(): Permite el ingreso o captura de la información (nombre, ubicación y estrellas) para ser almacenada en el objeto.

* mostrarDatos(): Despliega de forma organizada la información actual del hotel para su consulta por parte del usuario.

###3. Instanciación de Objetos
Para este modelo, se crean dos unidades específicas que representan hoteles reales en el sistema:

* hotel1: Objeto que almacenará la información del primer establecimiento capturado.

* hotel2: Objeto independiente que permite gestionar un segundo establecimiento de forma simultánea.

![texto del vínculo](infoclassxxx.png)

```python
class Hotel:
    def __init__(self):
        # Atributos inicializados vacíos
        self.nombre = ""
        self.ubicacion = ""
        self.estrellas = 0

    def leerDatos(self):
        print(f"\n--- 🏨 Registro de Nuevo Hotel ---")
        self.nombre = input("Nombre del hotel: ")
        self.ubicacion = input("Ubicación (Ciudad/Dirección): ")
        self.estrellas = input("Categoría (Número de estrellas): ")

    def mostrarDatos(self):
        print("\n-------------------------------")
        print(f"🏨 HOTEL: {self.nombre.upper()}")
        print(f"📍 Ubicación: {self.ubicacion}")
        print(f"⭐ Categoría: {self.estrellas} estrellas")
        print("-------------------------------")

# --- FLUJO INTERACTIVO CON 2 OBJETOS ---

# Creación del objeto hotel1
hotel1 = Hotel()
hotel1.leerDatos()

# Creación del objeto hotel2
hotel2 = Hotel()
hotel2.leerDatos()

# Mostrar los resultados finales
print("\n📋 RESUMEN DE HOTELES REGISTRADOS")
hotel1.mostrarDatos()
hotel2.mostrarDatos()
```
